# Executive Summary

## Olist E-Commerce Business Analysis

This project analyzes customer, order, revenue, delivery, payment, seller, product, and review data to identify the main business drivers, customer-experience risks, and opportunities for improvement.

The analysis combines descriptive analysis with cross-functional business analysis to translate operational and customer-level patterns into actionable recommendations.

## Executive Overview

The analysis shows a business with strong growth and a broad seller base, but with several important customer-experience and concentration risks.

Revenue growth has been accompanied primarily by increasing order volume rather than a meaningful increase in revenue per order. At the same time, repeat customers represent only a small share of the customer base despite having substantially higher average spend than one-time customers.

Delivery performance is one of the clearest customer-experience risks identified in the analysis. Late deliveries are strongly associated with lower review scores, making delivery reliability a high-priority operational area.

Revenue is also geographically concentrated, with a large share generated by a small number of customer states. This creates an opportunity to expand demand in underrepresented markets while maintaining the strength of the existing core markets.

Overall, the findings point toward three primary priorities: improving delivery reliability, increasing customer retention, and reducing key areas of customer and geographic concentration.

## Top Business Findings

### 1. Delivery Reliability Is a Major Customer-Experience Driver

Late deliveries are strongly associated with lower customer satisfaction. The average review score falls from 4.29 for on-time deliveries to 2.57 for late deliveries, making delivery reliability the clearest operational priority identified in the analysis.

### 2. Repeat Customers Represent a High-Value Opportunity

Repeat customers account for only 3.12% of the customer base, but their average spend is substantially higher than that of one-time customers. Increasing repeat purchases therefore represents an important opportunity to improve customer value.

### 3. Revenue Is Geographically Concentrated

São Paulo alone contributes 38.3% of total revenue, while the five largest customer states account for approximately 73.9%. This concentration creates an opportunity to diversify growth across additional markets.

### 4. Customer Satisfaction Varies Across Product Categories

Customer satisfaction is not uniform across product categories. `moveis_escritorio` has the lowest average review score among categories with at least 500 orders, indicating an area that warrants further investigation.

## Prioritized Recommendations

### 1. Improve Delivery Reliability — High Priority

Prioritize delivery reliability as an operational improvement area. Monitor late-delivery rates across key states, product categories, and sellers, and investigate the operational drivers behind persistent delays.

### 2. Strengthen Customer Retention — High Priority

Develop targeted retention initiatives focused on encouraging second purchases. Post-purchase engagement, personalized offers, and customer reactivation campaigns can be used to increase repeat purchasing.

### 3. Investigate Low-Performing Product Categories — Medium Priority

Start with `moveis_escritorio` and investigate product quality, fulfillment, seller performance, and customer complaints to identify the underlying drivers of lower satisfaction.

### 4. Reduce Geographic Revenue Concentration — Medium Priority

Expand strategically into underrepresented customer states by evaluating demand potential, seller coverage, and delivery capabilities while maintaining the strength of existing core markets.

## Supporting Opportunities

### Installment Payment Flexibility

Higher-value orders tend to use more payment installments. Maintaining flexible installment options may support higher-value purchases, although the current analysis does not establish a causal relationship between installment count and order value.

### Product Assortment and Pricing

Category revenue is strongly associated with sales volume, while the relationship between average price and revenue is considerably weaker. Product assortment and demand should therefore be considered alongside pricing when evaluating category growth opportunities.

## Limitations & Next Steps

### Limitations

- The analysis is primarily descriptive and identifies associations rather than causal relationships.
- Some findings are based on historical transaction data and may not represent current business conditions.
- Low-volume categories, sellers, and geographic markets may produce less reliable estimates and were filtered or interpreted cautiously where appropriate.
- The analysis identifies patterns associated with customer satisfaction but does not fully establish the operational causes of delivery delays or low category-level review scores.

### Next Steps

The highest-value next step would be to investigate the operational drivers behind late deliveries and low-performing product categories.

Further analysis could also evaluate customer retention strategies through controlled experiments, assess the profitability of installment-based purchasing, and identify specific geographic markets with the strongest potential for expansion.

These analyses would help validate the recommendations and move from descriptive findings toward measurable business interventions.
