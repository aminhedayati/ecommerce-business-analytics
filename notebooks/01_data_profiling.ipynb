# Data Profiling

This notebook explores the raw Olist e-commerce datasets to understand their structure, quality, and relationships before analysis.

```python
from pathlib import Path

import pandas as pd
```

```python
DATA_DIR = Path("../data/raw")
```

### Load Data

```python
files = sorted(DATA_DIR.glob("*.csv"))

files
```

```python
datasets = {
    file.stem: pd.read_csv(file)
    for file in files
}
```

```python
datasets["olist_orders_dataset"].head()
```

## Dataset Overview

The first step is to inspect the size of each dataset and understand the overall structure of the raw data.

```python
overview = pd.DataFrame(
    {
        "rows": {name: df.shape[0] for name, df in datasets.items()},
        "columns": {name: df.shape[1] for name, df in datasets.items()},
    }
)

overview
```

### Schema Inspection

To know what columns each table has and what their data type is.

```python
for name, df in datasets.items():
    print(f"\n{name}")
    display(
        pd.DataFrame({
            "column": df.columns,
            "dtype": df.dtypes.astype(str),
            "non_null": df.notna().sum().values,
            "null": df.isna().sum().values,
        })
    )
```

### Missing Values

```python
missing = []

for name, df in datasets.items():
    for column in df.columns:
        null_count = df[column].isna().sum()

        if null_count > 0:
            missing.append(
                {
                    "dataset": name,
                    "column": column,
                    "null_count": null_count,
                    "null_pct": round(null_count / len(df) * 100, 2),
                }
            )

missing_df = pd.DataFrame(missing)

missing_df.sort_values(
    ["dataset", "null_pct"],
    ascending=[True, False]
)
```

### Duplicate Records

Checking for complete duplicate rows.

```python
duplicates = pd.DataFrame(
    {
        "dataset": datasets.keys(),
        "duplicate_rows": [
            df.duplicated().sum()
            for df in datasets.values()
        ],
    }
)

duplicates
```

But we haven't decided yet which column is the Primary Key. We'll look at that in the next step.

### Key & Relationship Checks

Defining a list of expected keys.

```python
candidate_keys = {
    "olist_orders_dataset": ["order_id"],
    "olist_customers_dataset": ["customer_id"],
    "olist_products_dataset": ["product_id"],
    "olist_sellers_dataset": ["seller_id"],
}
```

Then checking:

```python
for dataset_name, columns in candidate_keys.items():
    df = datasets[dataset_name]

    for column in columns:
        print(
            dataset_name,
            column,
            "unique:", df[column].is_unique,
            "nulls:", df[column].isna().sum(),
        )
```

### Categorical Columns

Just a few important columns for now.

```python
categorical_checks = {
    "olist_orders_dataset": ["order_status"],
    "olist_order_payments_dataset": ["payment_type"],
    "olist_order_reviews_dataset": ["review_score"],
    "olist_products_dataset": ["product_category_name"],
}
```

```python
for dataset_name, columns in categorical_checks.items():
    df = datasets[dataset_name]

    for column in columns:
        print(f"\n{dataset_name} → {column}")
        display(df[column].value_counts(dropna=False))
```

### Date Columns

Identifing and converting timestamp columns.

```python
orders = datasets["olist_orders_dataset"].copy()

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for column in date_columns:
    orders[column] = pd.to_datetime(orders[column])
```

```python
orders[date_columns].agg(["min", "max"]).T
```

This gives us the time range of the data.

### Numeric Columns

At this point, just a preliminary look.

```python
for name, df in datasets.items():
    numeric_columns = df.select_dtypes(include="number").columns

    if len(numeric_columns) > 0:
        print(f"\n{name}")
        display(df[numeric_columns].describe().T)
```

We're not looking for outliers right now. We just want to know what the data looks like.
