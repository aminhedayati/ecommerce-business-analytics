# Analysis Setup

This notebook prepares the data structures and definitions used for downstream business analysis.

The focus is on preserving the correct grain of each dataset and avoiding incorrect aggregations caused by one-to-many relationships.

## Analysis Grains

The main analytical grains used in the project are:

- Order level: one row per order
- Order item level: one row per item within an order
- Customer level: one row per unique customer
- Seller level: one row per seller
- Product level: one row per product

## Core KPI Definitions

The following metrics will be used as the foundation for the business analysis.

Each metric is calculated at the grain appropriate to its business meaning.

### Orders

One order is counted once using `order_id`.

### Revenue

Revenue is calculated from order item prices using the `order_items` table.

### Customers

Customers are counted using `customer_unique_id`, which represents the actual customer across orders.

### Average Order Value

Average Order Value (AOV) is total revenue divided by the number of orders.

### Revenue Definition

Revenue is defined as the total value of products sold, calculated as the sum of `price` in the `order_items` table.

`freight_value` is treated separately as a shipping charge and is not included in revenue.

For analyses that require the total amount associated with an order, product revenue and freight charges can be combined.

## Baseline KPIs

The following metrics provide a high-level view of the marketplace before moving into more detailed business analysis.

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/raw")
```

```python
customers = pd.read_csv(DATA_DIR / "olist_customers_dataset.csv")
orders = pd.read_csv(DATA_DIR / "olist_orders_dataset.csv")
order_items = pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv")
payments = pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv")
products = pd.read_csv(DATA_DIR / "olist_products_dataset.csv")
sellers = pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv")
```

```python
total_orders = orders["order_id"].nunique()
total_revenue = order_items["price"].sum()
total_customers = customers["customer_unique_id"].nunique()
aov = total_revenue / total_orders   # average order value
```

```python
baseline_kpis = pd.DataFrame(
    {
        "metric": [
            "Orders",
            "Revenue",
            "Customers",
            "Average Order Value",
        ],
        "value": [
            f"{total_orders:,}",
            f"{total_revenue:,.2f}",
            f"{total_customers:,}",
            f"{aov:,.2f}",
        ],
    }
)

baseline_kpis
```
