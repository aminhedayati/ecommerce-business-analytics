# Business Recommendations

This notebook translates the key findings from the analysis into actionable business recommendations.

Each recommendation is linked to an observed business pattern and focuses on actions that could improve customer experience, retention, revenue resilience, or operational performance.

## Recommendation Framework

Recommendations are prioritized based on three factors:

- **Business impact:** the potential effect on revenue, customer experience, or operational performance.
- **Evidence strength:** how clearly the recommendation is supported by the analysis.
- **Actionability:** how directly the business could act on the finding.

## Recommendation Structure

Each recommendation follows a simple structure:

**Finding → Business Implication → Recommended Action**

The recommendations are based on the evidence identified in the previous analysis notebooks and do not imply causality where the analysis only establishes an association.

## Recommendation 1: Prioritize Delivery Reliability

### Finding

Late deliveries are strongly associated with lower customer satisfaction. Orders delivered after the estimated date received an average review score of 2.57, compared with 4.29 for on-time deliveries.

### Business Implication

Delivery reliability is an important customer-experience issue. Persistent delays can negatively affect customer satisfaction and may increase the risk of poor reviews and customer dissatisfaction.

### Recommended Action

Prioritize delivery reliability as an operational improvement area. Monitor late-delivery rates across key states, product categories, and sellers, and investigate the operational drivers behind persistent delays.

### Priority

**High**

## Recommendation 2: Strengthen Customer Retention

### Finding

Repeat customers represent only 3.12% of the customer base, but their average spend is substantially higher than that of one-time customers (259.87 vs. 137.63).

### Business Implication

The current customer base contains a relatively small group of repeat customers who generate substantially higher average spend. Increasing repeat purchasing could therefore represent an opportunity to improve customer value.

### Recommended Action

Develop targeted customer-retention initiatives focused on encouraging second purchases. Prioritize post-purchase engagement, personalized offers, and reactivation campaigns for customers who have already completed a purchase.

### Priority

**High**

## Recommendation 3: Investigate Low-Performing Product Categories

### Finding

Customer satisfaction varies meaningfully across product categories. `moveis_escritorio` has the lowest average review score among categories with at least 500 orders, at 3.64, while most other high-volume categories score above 4.0.

### Business Implication

Low customer satisfaction is not evenly distributed across the product portfolio. Categories with persistently lower review scores may represent specific areas of customer-experience risk.

### Recommended Action

Conduct a focused review of low-performing categories, starting with `moveis_escritorio`. Examine product quality, fulfillment, seller performance, and customer complaints within these categories to identify the underlying drivers of lower satisfaction.

### Priority

**Medium**

## Recommendation 4: Reduce Geographic Revenue Concentration

### Finding

Revenue is geographically concentrated, with São Paulo (SP) alone accounting for 38.3% of total revenue. The five largest customer states contribute approximately 73.9% of total revenue.

### Business Implication

A large share of revenue depends on a relatively small number of geographic markets. This concentration may increase exposure to regional demand changes and limits the diversification of the customer base.

### Recommended Action

Develop targeted growth strategies for underrepresented customer states. Prioritize markets with existing demand potential and evaluate localized marketing, seller coverage, and delivery capabilities before expanding aggressively.

### Priority

**Medium**

## Recommendation 5: Leverage Installment Flexibility for Higher-Value Purchases

### Finding

Higher-value orders tend to use more payment installments. The correlation between order value and the number of installments is 0.31, while average order value increases from 100.26 for one-installment orders to 377.66 for ten-installment orders.

### Business Implication

Installment-based payment flexibility is more common among higher-value purchases, suggesting that flexible payment options may be particularly relevant to customers making larger purchases.

### Recommended Action

Maintain and optimize installment payment options for higher-value purchases. Test whether targeted installment messaging or promotional offers can improve conversion or average order value, while monitoring payment costs and profitability.

### Priority

**Low**

## Recommendation 6: Optimize Product Assortment and Pricing

### Finding

Revenue varies substantially across product categories and is strongly associated with sales volume. The correlation between items sold and revenue is 0.95, while the relationship between average price and revenue is considerably weaker.

### Business Implication

Category revenue appears to be driven primarily by sales volume rather than price alone. This suggests that increasing revenue through price changes without considering demand and sales volume may not be effective across all categories.

### Recommended Action

Prioritize assortment and demand when evaluating category performance. Identify categories with strong sales volume but lower revenue potential for targeted pricing or product-mix optimization, while avoiding broad price increases that could reduce demand.

### Priority

**Low**

# Recommendation Prioritization

The recommendations are prioritized based on their expected business impact, strength of supporting evidence, and practical actionability.

| Investigate low-performing product categories | High | Moderate | Medium | 3 |
| Reduce geographic revenue concentration | Medium | Strong | Medium | 4 |

## Executive Priority

The highest-priority actions are improving delivery reliability and strengthening customer retention. Both areas combine meaningful business impact with relatively strong evidence and clear opportunities for action.

Product-category performance and geographic concentration represent secondary priorities that warrant targeted investigation and strategic planning.

Payment installments and assortment/pricing should be treated as supporting opportunities rather than immediate strategic priorities, as the current analysis identifies associations but does not establish causal effects.
